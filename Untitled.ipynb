```python
KEY = "4d4163717179757236366454526347"

AREA_NM = "POI013"

url = f"http://openapi.seoul.go.kr:8088/{KEY}/xml/citydata/1/10/{AREA_NM}"
```

```python
response = requests.get(url, timeout=30)

print("응답 상태 코드:", response.status_code)

if response.status_code == 200:
    print(response.text[:2000])
else:
    print("API 요청 실패")
```

```python
def get_population_data():

    response = requests.get(url, timeout=30)
    response.raise_for_status()

    root = ET.fromstring(response.text)

    result_code = root.findtext(".//RESULT/CODE")

    if result_code and result_code != "INFO-000":
        result_msg = root.findtext(".//RESULT/MESSAGE")
        raise Exception(f"API 오류: {result_code} - {result_msg}")

    def get_text(tag):
        element = root.find(f".//{tag}")
        if element is not None and element.text:
            return element.text.strip()
        return None

    data = {
        "수집시간": datetime.now().strftime("%Y-%m-%d %H:%M:%S"),
        "장소명": get_text("AREA_NM"),
        "혼잡도": get_text("AREA_CONGEST_LVL"),
        "혼잡도메시지": get_text("AREA_CONGEST_MSG"),
        "최소인구": get_text("AREA_PPLTN_MIN"),
        "최대인구": get_text("AREA_PPLTN_MAX"),
        "남성비율": get_text("MALE_PPLTN_RATE"),
        "여성비율": get_text("FEMALE_PPLTN_RATE"),
        "데이터기준시간": get_text("PPLTN_TIME")
    }

    return data
```

```python
test_data = get_population_data()

print(test_data)
```

```python
records = []

for i in range(19):

    print(f"{i + 1}/19 번째 데이터 수집 중...")

    try:
        data = get_population_data()
        records.append(data)

        print(data)

        pd.DataFrame(records).to_csv(
            "gasan_population.csv",
            index=False,
            encoding="utf-8-sig"
        )

    except Exception as e:
        print("수집 오류:", e)

    if i < 18:
        print("10분 후 다음 데이터를 수집")
        time.sleep(600)

print("데이터 수집 완료")
```

```python
df = pd.DataFrame(records)

df["최소인구"] = pd.to_numeric(df["최소인구"], errors="coerce")
df["최대인구"] = pd.to_numeric(df["최대인구"], errors="coerce")
df["남성비율"] = pd.to_numeric(df["남성비율"], errors="coerce")
df["여성비율"] = pd.to_numeric(df["여성비율"], errors="coerce")

df.to_csv(
    "gasan_population.csv",
    index=False,
    encoding="utf-8-sig"
)

display(df)

display(FileLink("gasan_population.csv"))
```

```python
plt.figure(figsize=(12, 5))

plt.plot(df["수집시간"], df["최소인구"],
         marker="o", label="Minimum Population")

plt.plot(df["수집시간"], df["최대인구"],
         marker="o", label="Maximum Population")

plt.title("Gasan Digital Complex Station - Population")
plt.xlabel("Collection Time")
plt.ylabel("Population")

plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.tight_layout()
plt.show()
```
